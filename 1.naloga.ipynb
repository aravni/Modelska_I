```python
import numpy as np
import sympy as sym
from matplotlib import pyplot as plt
import matplotlib.patches as mpatches
from labellines import labelLine, labelLines
```

```python
def ekstremna_v(t, v_0, t_0, l):
    return 3*(l-v_0*t_0)/(2*t_0**3)*(2*t_0*t - t**2) + v_0 #a(t_0) = 0
def max_v(t, v_0, t_0, l, v_max):
    return 3*(2*l-v_0*t_0-v_max*t_0)/t_0**3*(t_0*t-t**2) + (v_max-v_0)/t_0*t + v_0 #v(t_0) = v_max
def zač_pospešek(t, v_0, t_0, l):
    return -6/t_0**3 * (l-v_0*t_0) * t**2 /4 + v_0 #a(0)=0
```

```python
v_0 = 20 #m/s
t_0 = 10 #s
l = 300 #m

t = np.linspace(0, t_0)
v = ekstremna_v(t, v_0, t_0, l)
v1 = zač_pospešek(t, v_0, t_0, l)

plt.plot(t, v)
plt.plot(t, v1)
np.trapz(v, t), np.trapz(v1, t)
```

```python
v_0 = 0/3.6 #km/h -> m/s
t_0 = 60 #s
l = 750 #m
v_max = 50/3.6 #km/h -> m/s

t1 = np.linspace(0, t_0)
v_1 = ekstremna_v(t1, v_0, t_0, l)
v_2 = max_v(t1, v_0, t_0, l, v_max)

#t_0 = 10 #s
l = 500
t2 = np.linspace(0, t_0)
v_3 = ekstremna_v(t2, v_0, t_0, l)
v_4 = max_v(t2, v_0, t_0, l, v_max)

#t_0 = 20 #s
l = 1000
t3 = np.linspace(0, t_0)
v_5 = ekstremna_v(t3, v_0, t_0, l)
v_6 = max_v(t3, v_0, t_0, l, v_max)

plt.plot(t1, v_1, 'C0', label='0,75 km')
plt.plot(t1, v_2, 'C1', label='0,75 km')
plt.plot(t2, v_3, 'C0', label='0,5 km')
plt.plot(t2, v_4, 'C1', label='0,5 km')
plt.plot(t3, v_5, 'C0', label='1 km')
plt.plot(t3, v_6, 'C1', label='1 km')
labelLines(plt.gca().get_lines())
b1 = mpatches.Patch(color='C0', label='Ekstremna končna hitrost')
b2 = mpatches.Patch(color='C1', label='Končna hitrost $v_{max} = 50$ km/h')
plt.legend(handles=[b1, b2])
plt.plot(0,0)
plt.xlabel('t [s]')
plt.ylabel('v [m/s]')
#plt.savefig("C:/Users/ravni/Namizje/0_do_50kmh.pdf", bbox_inches='tight')
plt.show()
np.trapz(v_1, t1), np.trapz(v_2, t1)
```
